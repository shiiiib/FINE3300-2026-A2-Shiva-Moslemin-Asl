```python
import pandas as pd
import numpy as np


class InflationCalculator:

    def __init__(self, filename):
        self.data = pd.read_csv(filename, skiprows=8)
        self.data.columns = ["Date", "CPI"]
        self.dates = list(self.data["Date"])

    def get_price(self):
        while True:
            try:
                entry = input("Enter the price in CAD (must be > 0, can include decimals): ")
                price = float(entry.replace(",", "").replace(" ", ""))

                if price > 0 and np.isfinite(price): 
                    return price

                print("Enter a positive, finite price.")

            except ValueError:
                print("Enter a valid number.")

    def get_date(self, prompt):
        date = input(prompt).strip()

        while date + "-01" not in self.dates:
            print("Enter an available month in YYYY-MM format.")
            date = input(prompt).strip()

        return date

    def get_cpi(self, date):
        row = self.data[self.data["Date"] == date + "-01"]
        return row["CPI"].iloc[0]

    def calculate(self):
        # Get the user's inputs.
        price = self.get_price()
        first_date = self.get_date("Enter the first month (YYYY-MM): ")
        second_date = self.get_date("Enter the second month (YYYY-MM): ")

        first_cpi = self.get_cpi(first_date)
        second_cpi = self.get_cpi(second_date)

        # Calculate the price and percentage change.
        ratio = second_cpi / first_cpi
        new_price = price * ratio
        percent_change = (ratio - 1) * 100

        # Extract the years and months.
        first_year = int(first_date[:4])
        second_year = int(second_date[:4])
        first_month = int(first_date[5:7])
        second_month = int(second_date[5:7])

        months = (second_year - first_year) * 12 + second_month - first_month # Calculate the difference between the dates in months.
        years = abs(months) / 12 # Convert the month difference into a positive number of years.

        # Display the results.
        print("\nCANADIAN INFLATION CALCULATOR")
        print(f"Original price in {first_date}: CAD ${price:,.2f}")
        print(f"Equivalent price in {second_date}: CAD ${new_price:,.2f}")
        print(f"Total percentage change: {percent_change:.2f}%")
        print(f"Number of years: {years:.2f}")
        print(f"CPI for {first_date}: {first_cpi:.1f}")
        print(f"CPI for {second_date}: {second_cpi:.1f}")

        if years == 0:
            print("Average annual inflation rate: N/A (same month)")
        else:
            annual_rate = (ratio ** (1 / years) - 1) * 100
            print(f"Average annual inflation rate: {annual_rate:.2f}%")


calculator = InflationCalculator("cpi_data.csv")
calculator.calculate()
```
